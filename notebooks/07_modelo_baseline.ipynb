## Objetivo del notebook

Establecer la línea base contra la que se comparará el rendimiento de los modelos avanzados (Prophet y LightGBM). El baseline se define como la **media móvil de los precios de los últimos 7 días por estación**, una regla aritmética que no requiere entrenamiento.

## Razón de incluir un baseline

1. **Validez metodológica**: cualquier modelo complejo debe demostrar mejora cuantificable frente a una regla simple.
2. **Series temporales muy inerciales**: en mercados como el de carburantes, donde los precios cambian poco día a día, el baseline puede ser sorprendentemente robusto.

## Métricas de evaluación

- **MAE** (Mean Absolute Error): expresado en céntimos por litro, interpretable directamente.
- **RMSE** (Root Mean Squared Error): penaliza errores extremos.
- **MAPE** (Mean Absolute Percentage Error): porcentaje de error medio.

## Combinaciones evaluadas

12 evaluaciones en total: 2 carburantes × 2 régimenes × 3 conjuntos (train, val, test).

```python
# Configuración del entorno
import sys
import warnings
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["figure.figsize"] = (14, 6)
plt.rcParams["font.family"] = "Calibri"
plt.rcParams["font.size"] = 11
sns.set_style("whitegrid")
warnings.filterwarnings("ignore", category=FutureWarning)

# Añadir raíz del proyecto al sys.path
RAIZ_PROYECTO = Path("..").resolve()
if str(RAIZ_PROYECTO) not in sys.path:
    sys.path.insert(0, str(RAIZ_PROYECTO))

# Importar funciones del módulo de modelos
from src.modelos import (evaluar,imprimir_evaluacion,cargar_todas_particiones,predecir_baseline)

# Rutas
CARPETA_PARTICIONES = Path("../data/processed/particiones")
CARPETA_RESULTADOS = Path("../outputs/resultados_modelos")
CARPETA_RESULTADOS.mkdir(parents=True, exist_ok=True)

print(f"Raíz del proyecto: {RAIZ_PROYECTO}")
print(f"Carpeta particiones: {CARPETA_PARTICIONES.resolve()}")
print(f"Carpeta resultados:  {CARPETA_RESULTADOS.resolve()}")
print("\nMódulos importados correctamente")
```

## Evaluación del baseline: Gasóleo A · Régimen pre-shock

Cargamos las particiones del Gasóleo A en régimen pre-shock y evaluamos el baseline en los tres conjuntos.

```python
# Gasóleo A · Pre-shock
print(">>> Gasóleo A · Pre-shock")
print("=" * 70)

particiones = cargar_todas_particiones(CARPETA_PARTICIONES, "gasoleo_a", "pre_shock")

resultados_gasoleo_pre = []

for conjunto in ["train", "val", "test"]:
    df = particiones[conjunto]
    y_real = df["precio"]
    y_predicho = predecir_baseline(df)
    
    resultado = evaluar(y_real, y_predicho, 
                        nombre=f"Gasóleo A · Pre-shock · {conjunto}")
    resultados_gasoleo_pre.append(resultado)

imprimir_evaluacion(resultados_gasoleo_pre)
```

## Evaluación del baseline: Gasóleo A · Régimen post-shock

Repetimos la evaluación para el régimen post-shock.

```python
# Gasóleo A · Post-shock
print(">>> Gasóleo A · Post-shock")
print("=" * 70)

particiones = cargar_todas_particiones(CARPETA_PARTICIONES, "gasoleo_a", "post_shock")

resultados_gasoleo_post = []

for conjunto in ["train", "val", "test"]:
    df = particiones[conjunto]
    y_real = df["precio"]
    y_predicho = predecir_baseline(df)
    
    resultado = evaluar(y_real, y_predicho, 
                        nombre=f"Gasóleo A · Post-shock · {conjunto}")
    resultados_gasoleo_post.append(resultado)

imprimir_evaluacion(resultados_gasoleo_post)
```

## Evaluación del baseline: Gasolina 95 E5 · Ambos régimenes

```python
# Gasolina 95 E5 · Pre-shock
print(">>> Gasolina 95 E5 · Pre-shock")
print("=" * 70)

particiones = cargar_todas_particiones(CARPETA_PARTICIONES, "gasolina_95", "pre_shock")

resultados_gasolina_pre = []

for conjunto in ["train", "val", "test"]:
    df = particiones[conjunto]
    y_real = df["precio"]
    y_predicho = predecir_baseline(df)
    
    resultado = evaluar(y_real, y_predicho, 
                        nombre=f"Gasolina 95 · Pre-shock · {conjunto}")
    resultados_gasolina_pre.append(resultado)

imprimir_evaluacion(resultados_gasolina_pre)

# Gasolina 95 E5 · Post-shock
print("\n\n>>> Gasolina 95 E5 · Post-shock")
print("=" * 70)

particiones = cargar_todas_particiones(CARPETA_PARTICIONES, "gasolina_95", "post_shock")

resultados_gasolina_post = []

for conjunto in ["train", "val", "test"]:
    df = particiones[conjunto]
    y_real = df["precio"]
    y_predicho = predecir_baseline(df)
    
    resultado = evaluar(y_real, y_predicho, 
                        nombre=f"Gasolina 95 · Post-shock · {conjunto}")
    resultados_gasolina_post.append(resultado)

imprimir_evaluacion(resultados_gasolina_post)
```

## Consolidación de resultados y guardado

Reunimos los resultados de las 12 evaluaciones en una tabla única y la guardamos como CSV para que los modelos posteriores (Prophet, LightGBM) puedan compararse fácilmente.

```python
# Consolidar todos los resultados
todos_resultados = (resultados_gasoleo_pre + resultados_gasoleo_post +resultados_gasolina_pre + resultados_gasolina_post)

# Convertir a DataFrame
df_resultados = pd.DataFrame(todos_resultados)
df_resultados["modelo"] = "baseline_mm7"

# Reordenar columnas
df_resultados = df_resultados[["modelo", "nombre", "n_observaciones", "mae_cts", "rmse_cts", "mape_pct"]]

# Guardar
ruta_resultados = CARPETA_RESULTADOS / "baseline_resultados.csv"
df_resultados.to_csv(ruta_resultados, index=False)

print(f"=" * 70)
print(f"RESUMEN GLOBAL DEL BASELINE")
print(f"=" * 70)
print(df_resultados.to_string(index=False))
print(f"\nResultados guardados en: {ruta_resultados}")
print(f"  ({len(df_resultados)} filas)")

# Resumen ejecutivo
print(f"\n{'─' * 70}")
print(f"RESUMEN EJECUTIVO")
print(f"{'─' * 70}")
print(f"\nMAE promedio en TEST por carburante × régimen:")
test_only = df_resultados[df_resultados["nombre"].str.contains("test")]
print(test_only[["nombre", "mae_cts", "mape_pct"]].to_string(index=False))
```
